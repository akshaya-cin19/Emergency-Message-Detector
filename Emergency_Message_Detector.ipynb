```python
# Emergency Message Detector

import re

messages = [
    "Help! There is a fire in the building. Please send an ambulance immediately.",
    "I will reach college by 10 AM tomorrow.",
    "Urgent! My friend met with an accident and needs medical help.",
    "Can you send me the project report today?",
    "Emergency! Someone is trapped inside the building. Call the fire department."
]

emergency_keywords = {
    "fire": 5,
    "accident": 5,
    "ambulance": 5,
    "emergency": 5,
    "help": 4,
    "urgent": 4,
    "trapped": 5,
    "danger": 4,
    "injured": 5,
    "medical": 4,
    "police": 5,
    "rescue": 5,
    "fire department": 5
}

def analyze_message(message):

    text = message.lower()
    score = 0
    detected = []

    for keyword, weight in emergency_keywords.items():
        if keyword in text:
            score += weight
            detected.append(keyword)

    if score >= 10:
        status = "HIGH EMERGENCY"
    elif score >= 5:
        status = "MEDIUM EMERGENCY"
    else:
        status = "NORMAL"

    return score, status, detected


print("=" * 60)
print("             EMERGENCY MESSAGE DETECTOR")
print("=" * 60)

for i, message in enumerate(messages, 1):

    score, status, detected = analyze_message(message)

    print(f"\nMessage {i}:")
    print(message)

    print("Emergency Score :", score)
    print("Status          :", status)

    if detected:
        print("Detected Words  :", ", ".join(detected))
    else:
        print("Detected Words  : None")

    print("-" * 60)

print("\nAnalysis Completed Successfully!")
```
